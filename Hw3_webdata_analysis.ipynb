```python
#Configure function to pull based on the dictionary for the ETFs
def pull_source_data(series_id, series_name):
    """ Reading the data series from FRED """
    ##configure url
    url = f"https://fred.stlouisfed.org/graph/fredgraph.csv?id={series_id}"

    ##read the data
    source_df = pd.read_csv(url)

    ##rename the columns for visualization
    source_df = source_df.rename(columns={"observation_date":"date"})
    source_df = source_df.rename(columns={series_id:"value"})

    ##adding the columns 
    source_df["series_id"] = series_id
    source_df["series_name"] = series_name

    return source_df
```
