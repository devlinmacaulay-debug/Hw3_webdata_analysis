```python
#Build a dictionary to feed a function to pull S&P 500, Nasdaq, VIX and the Dow - using the id code
source_data_dict = {"SP500":"S&P 500", "NASDAQCOM":"NASDAQ", "VIXCLS":"VIX", "DJIA":"Dow Jones"}

source_data_dict
```

```python
#Configure function to pull based on the dictionary for the ETFs
def pull_source_data(series_id, series_name):
    """ Reading the data series from FRED """
    ##configure url
    url = f"https://fred.stlouisfed.org/graph/fredgraph.csv?id={series_id}"

    ##read the data
    source_df = pd.read_csv(url)

    ##rename the columns for visualization
    source_df = source_df.rename(columns={"observation_date":"date"})
    source_df = source_df.rename(columns={series_id:"value"})

    ##adding the columns 
    source_df["series_id"] = series_id
    source_df["series_name"] = series_name

    return source_df
```

After pulling the Data from FRED on SP500, NASDAW, VIX, and Dow Jones and testing the data pulls, I will be adding all the data to one data frame to create visualizations in the future.

```python
#Creating df to store data from all 4 ETFs

##Configure empty list variable
tot_source_data_list = []

##Looping through each of the source data pulls in the dictionary
for series_id, series_name in source_data_dict.items():
    ##Pulling data source
    single_source_df = pull_source_data(series_id, series_name)

    ##Add it to the list
    tot_source_data_list.append(single_source_df)

##Combine everything 
tot_market_df = pd.concat(tot_source_data_list, ignore_index=True)

##Review the results top
tot_market_df.head(10)
```
